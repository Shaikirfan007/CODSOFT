```python
# Question 7: Write a program to reverse a given integer number.

num = int(input())

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num = num // 10

print(rev)
```

```python
# Question 8: Write a program to check whether a given number is a palindrome.

num = int(input())

original = num
reverse = 0

while num > 0:
    digit = num % 10
    reverse = reverse * 10 + digit
    num = num // 10

if original == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# Question 44: Print all prime numbers up to N using a function.

def isPrime(num):
    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True

n = int(input("Enter N: "))

for i in range(2, n + 1):
    if isPrime(i):
        print(i, end=" ")
```

```python
# Question 45: Recursive factorial.

def factorial(n):
    if n == 0 or n == 1:
        return 1

    return n * factorial(n - 1)

num = int(input("Enter a number: "))

print("Factorial =", factorial(num))
```

```python
# Question 46: Recursive Fibonacci.

def fibonacci(n):
    if n == 0:
        return 0

    if n == 1:
        return 1

    return fibonacci(n - 1) + fibonacci(n - 2)

n = int(input("Enter N: "))

print("Nth Fibonacci Number =", fibonacci(n))
```

```python
# Question 47: Linear Search

numbers = list(map(int, input("Enter list elements: ").split()))

key = int(input("Enter element to search: "))

found = False

for i in range(len(numbers)):
    if numbers[i] == key:
        print("Element found at index", i)
        found = True
        break

if found == False:
    print("Element not found")
```

```python
# Question 48: Binary Search

numbers = list(map(int, input("Enter sorted list: ").split()))

key = int(input("Enter element to search: "))

low = 0
high = len(numbers) - 1

while low <= high:
    mid = (low + high) // 2

    if numbers[mid] == key:
        print("Element found at index", mid)
        break

    elif numbers[mid] < key:
        low = mid + 1

    else:
        high = mid - 1

else:
    print("Element not found")
```

```python
# Question 49: Bubble Sort

numbers = list(map(int, input("Enter list elements: ").split()))

n = len(numbers)

for i in range(n):
    for j in range(0, n - i - 1):
        if numbers[j] > numbers[j + 1]:
            numbers[j], numbers[j + 1] = numbers[j + 1], numbers[j]

print("Sorted List:", numbers)
```

```python
# Question 50: Selection Sort

numbers = list(map(int, input("Enter list elements: ").split()))

n = len(numbers)

for i in range(n):
    min_index = i

    for j in range(i + 1, n):
        if numbers[j] < numbers[min_index]:
            min_index = j

    numbers[i], numbers[min_index] = numbers[min_index], numbers[i]

print("Sorted List:", numbers)
```
